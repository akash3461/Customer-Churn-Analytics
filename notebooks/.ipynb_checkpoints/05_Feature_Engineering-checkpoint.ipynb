```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import joblib

# Load clean data
df = pd.read_csv("../data/processed/customer_churn_clean.csv")

# Confirm dynamic column names
target_col = 'churn_label' if 'churn_label' in df.columns else 'churn'
tenure_col = 'tenure_months' if 'tenure_months' in df.columns else 'tenure'

print(f"Data loaded: {df.shape}")
```

### 1. Feature Creation
Building quick practical features based on customer behavior:
* `tenure_cohort`: Grouping tenure into standard customer lifecycle buckets (0-1 yr, 1-2 yrs, etc.).
* `total_services`: Counting active add-ons (security, backup, protection, streaming, support).
* `avg_monthly_spend`: Historical average charge per month (`total_charges / tenure`).

```python
df_fe = df.copy()

# 1. Tenure Cohorts
bins = [-1, 12, 24, 48, 60, np.inf]
labels = ['0-1 yr', '1-2 yrs', '2-4 yrs', '4-5 yrs', '5+ yrs']
df_fe['tenure_cohort'] = pd.cut(df_fe[tenure_col], bins=bins, labels=labels)

# 2. Total Services Subscribed
service_cols = ['online_security', 'online_backup', 'device_protection', 
                'tech_support', 'streaming_tv', 'streaming_movies']

existing_services = [c for c in service_cols if c in df_fe.columns]
df_fe['total_services'] = (df_fe[existing_services] == 'Yes').sum(axis=1)

# 3. Average Monthly Spend Ratio (safely avoid div by 0)
df_fe['avg_monthly_spend'] = np.where(
    df_fe[tenure_col] > 0, 
    df_fe['total_charges'] / df_fe[tenure_col], 
    df_fe['monthly_charges']
)

df_fe[['total_services', 'avg_monthly_spend', 'tenure_cohort']].head()
```

### 2. Binary Target Encoding & Feature Selection
Dropping non-predictive identifiers (`customer_id` or similar) and mapping `churn` to 0/1.

```python
# Map target column to binary (1 = Churned, 0 = Retained)
if df_fe[target_col].dtype == 'object':
    df_fe['target'] = df_fe[target_col].map({'Yes': 1, 'No': 0})
else:
    df_fe['target'] = df_fe[target_col]

# Drop ID columns, raw targets, or high-cardinality metadata
drop_cols = [target_col, 'target']
for col in ['customer_id', 'customer_id_label', 'country', 'state', 'city', 'zip_code', 'lat_long', 'latitude', 'longitude']:
    if col in df_fe.columns:
        drop_cols.append(col)

X = df_fe.drop(columns=drop_cols)
y = df_fe['target']

print(f"Features matrix shape: {X.shape}")
print(f"Target distribution:\n{y.value_counts(normalize=True).round(3)}")
```

### 3. Categorical Encoding & Train/Test Split
Using One-Hot Encoding (`pd.get_dummies`) on nominal variables and splitting with stratified sampling to preserve class distribution.

```python
# One-hot encode categorical features
X_encoded = pd.get_dummies(X, drop_first=True)

# 80/20 Train-Test split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Train shape: {X_train.shape}")
print(f"Test shape:  {X_test.shape}")
```

### 4. Feature Scaling & Saving
Applying `StandardScaler` to train and test sets to prevent data leakage, then saving preprocessor for model building.

```python
scaler = StandardScaler()

# Fit only on train data, transform both
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns)

# Save preprocessor artifacts
joblib.dump(scaler, "../models/preprocessor.pkl")

print("✅ Feature engineering complete and scaler saved to models/preprocessor.pkl!")
```
